# 01 — Data Download & Preprocessing

Downloads weekly adjusted close prices for all current S&P 500 constituents via yfinance,
computes log returns, subtracts the risk-free rate, and saves the excess return matrix to disk.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
from src.data.download import get_sp500_tickers, download_prices
from src.data.preprocess import to_log_returns, get_risk_free_rate, to_excess_returns
```

## 1. Download prices

```python
START = "2020-01-01"
END   = "2025-04-01"

tickers = get_sp500_tickers()
print(f"Fetched {len(tickers)} tickers")

prices = download_prices(tickers, start=START, end=END, interval="1wk")
print(f"Price matrix: {prices.shape}  ({prices.shape[0]} weeks × {prices.shape[1]} stocks)")
prices.tail(3)
```

## 2. Log returns & excess returns

```python
# 4.22% annual risk-free rate (approximate 2024 T-bill average)
RF_ANNUAL = 0.0422

log_ret = to_log_returns(prices)
rf      = get_risk_free_rate(annual_rate=RF_ANNUAL, freq="W")
excess  = to_excess_returns(log_ret, rf)

print(f"Excess return matrix: {excess.shape}")
excess.describe().T[["mean", "std", "min", "max"]].head(10)
```

## 3. Save to disk

```python
prices.to_csv("../data/prices_weekly.csv")
excess.to_csv("../data/excess_returns_weekly.csv")
print("Saved prices_weekly.csv and excess_returns_weekly.csv")
```
